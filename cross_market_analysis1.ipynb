```python
import requests
import time

all_records = []

# Loop through pages 1 to 5
for page_number in range(1, 6):
    # Use f-string to insert the page_number into the URL
    url = f"https://api.coingecko.com/api/v3/coins/markets?vs_currency=inr&per_page=250&order=market_cap_desc&page={page_number}&sparkline=False"

    response = requests.get(url)

    if response.status_code == 200:
        data = response.json()  # Convert the response to a list of dictionaries
        all_records.extend(data) # Add the new list of coins to our main list
        print(f"Page {page_number} collected. Total coins: {len(all_records)}")
    else:
        print(f"Failed to fetch Page {page_number}. Error: {response.status_code}")
        break

    # Pro-tip: CoinGecko has rate limits. Adding a tiny sleep prevents getting blocked.
    time.sleep(1)

print(f"Successfully collected {len(all_records)} coin records!")
```

```python
records = []
for i in all_records:
    records.append({
        'id': i['id'], 
        'symbol': i['symbol'], 
        'name': i['name'], 
        'current_price': i['current_price'], 
        'market_cap': i['market_cap'], 
        'market_cap_rank': i['market_cap_rank'], 
        'total_volume': i['total_volume'],
        'circulating_supply': i['circulating_supply'], 
        'total_supply': i['total_supply'], 
        'ath': i['ath'], 
        'atl': i['atl'], 
        'last_updated': i['last_updated']
    })
records
```

```python
import pandas as pd
coin_df = pd.DataFrame(records)
coin_df
```

```python
coin_df['last_updated'] = pd.to_datetime(coin_df['last_updated']).dt.date
coin_df = coin_df.rename(columns={'last_updated': 'date'})
coin_df
```

```python
coin_top = []
coin_id = ['bitcoin', 'ethereum', 'tether', 'binancecoin', 'ripple']

for i in coin_id:
  url = f"https://api.coingecko.com/api/v3/coins/{i}/market_chart?vs_currency=usd&days=365"

  response = requests.get(url)
  
  if response.status_code == 200:
      data = response.json()
      print(data)
      for item in data['prices']:
        coin_top.append({
            'id': i,
            'date': item[0],
            'price': item[1]
        })
      print(f"Successfully fetched data for {i}")
      time.sleep(10)
  else:
        print(f"Failed to fetch data for {i}. Error: {response.status_code}")
        break

print(coin_top)
```

```python
coin_top_df = pd.DataFrame(coin_top)
coin_top_df
```

```python
# Convert Unix Milliseconds to readable YYYY-MM-DD
coin_top_df['date'] = pd.to_datetime(coin_top_df['date'], unit='ms').dt.date
coin_top_df = coin_top_df.rename(columns={'id': 'coin_id', 'price': 'price_usd'})
coin_top_df 
```

```python
import pymysql

# ---------------------------------------------------------
# 1. CONFIGURATION
# ---------------------------------------------------------
DB_CONFIG = {
    'host': 'localhost',
    'user': 'root',
    'password': 'Suba@2000',
    # NOTE: Do NOT include 'database' here yet!
}
DB_NAME = 'test_analysis1'  # Matches your error message

def create_schema():
    print("🔌 Connecting to MySQL...")
    # 1. Connect to MySQL Server (No specific DB)
    conn = pymysql.connect(**DB_CONFIG)
    cursor = conn.cursor()

    # 2. Drop Database if it exists (Fresh Start)
    print(f"🗑️  Dropping Database '{DB_NAME}' if it exists...")
    cursor.execute(f"DROP DATABASE IF EXISTS {DB_NAME}")

    # 3. Create Database
    print(f"⚙️  Creating Database '{DB_NAME}'...")
    cursor.execute(f"CREATE DATABASE {DB_NAME}")
    print("✅ Database created!")

    # 4. Select the new Database
    conn.select_db(DB_NAME)

    # 5. Create Tables
    print("📝 Creating Tables...")

    # Table 1: Cryptocurrencies
    cursor.execute("""
    CREATE TABLE Cryptocurrencies (
        id VARCHAR(250) PRIMARY KEY,
        symbol VARCHAR(250),
        name VARCHAR(250),
        current_price DOUBLE,
        market_cap DOUBLE,
        market_cap_rank INT,
        total_volume DOUBLE,
        circulating_supply DOUBLE,
        total_supply DOUBLE,
        ath DOUBLE,
        atl DOUBLE,
        date DATE
    )
    """)
    print("   ✅ Cryptocurrencies created")

    # Table 2: Crypto_prices
    cursor.execute("""
    CREATE TABLE Crypto_prices (
        coin_id VARCHAR(250),
        date DATE,
        price_usd DOUBLE,
        FOREIGN KEY (coin_id) REFERENCES Cryptocurrencies(id)
    )
    """)
    print("   ✅ Crypto_prices created")

    # Table 3: Oil_prices
    cursor.execute("""
    CREATE TABLE Oil_prices (
        date DATE PRIMARY KEY,
        price_usd DECIMAL(18, 6)
    )
    """)
    print("   ✅ Oil_prices created")

    # Table 4: Stock_prices
    cursor.execute("""
    CREATE TABLE Stock_prices (
        date DATE,
        open DECIMAL(18, 6),
        high DECIMAL(18, 6),
        low DECIMAL(18, 6),
        close DECIMAL(18, 6),
        volume BIGINT,
        ticker VARCHAR(20)
    )
    """)
    print("   ✅ Stock_prices created")

    conn.close()
    print("\n🎉 Database and Tables created successfully!")

if __name__ == "__main__":
    create_schema()
```

```python
import pymysql
import pandas as pd

# 1. Connect to MySQL
connection = pymysql.connect(
    host='localhost',
    user='root',
    password='Suba@2000',
    database='test_analysis1'
)
cursor = connection.cursor()
print("Connected to database successfully!")

# ---------------------------------------------------------
# 1. UPLOAD CRYPTOCURRENCIES
# ---------------------------------------------------------
print("1. Uploading Cryptocurrencies (Row by Row)...")

if 'last_updated' in coin_df.columns:
    coin_df = coin_df.rename(columns={'last_updated': 'date'})

columns_needed = ['id', 'symbol', 'name', 'current_price', 'market_cap', 
                  'market_cap_rank', 'total_volume', 'circulating_supply', 
                  'total_supply', 'ath', 'atl', 'date']
df_to_upload = coin_df[columns_needed].copy()
df_to_upload = df_to_upload.where(pd.notnull(df_to_upload), None)

# SQL Command (INSERT IGNORE handles duplicates)
sql = """
INSERT IGNORE INTO Cryptocurrencies 
(id, symbol, name, current_price, market_cap, market_cap_rank, 
 total_volume, circulating_supply, total_supply, ath, atl, date) 
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

# Loop through each row and execute
data_list = [tuple(x) for x in df_to_upload.to_numpy()]
for row in data_list:
    cursor.execute(sql, row)

connection.commit()
print(f"   Done! ({len(data_list)} rows processed)")


# ---------------------------------------------------------
# 2. UPLOAD CRYPTO PRICES
# ---------------------------------------------------------
print("2. Uploading Crypto Prices (Row by Row)...")

prices_df = coin_top_df.rename(columns={'id': 'coin_id', 'price': 'price_usd'})
prices_df['date'] = pd.to_datetime(prices_df['date']).dt.date
prices_df = prices_df[['coin_id', 'date', 'price_usd']]
prices_df = prices_df.where(pd.notnull(prices_df), None)

sql = "INSERT IGNORE INTO Crypto_prices (coin_id, date, price_usd) VALUES (%s, %s, %s)"

data_list = [tuple(x) for x in prices_df.to_numpy()]
for row in data_list:
    cursor.execute(sql, row)

connection.commit()
print(f"   Done! ({len(data_list)} rows processed)")


# ---------------------------------------------------------
# 3. UPLOAD OIL PRICES
# ---------------------------------------------------------
print("3. Uploading Oil Prices (Row by Row)...")

oil_df = filtered_oil_df.rename(columns={'Date': 'date', 'Price': 'price_usd'})
oil_df['date'] = pd.to_datetime(oil_df['date']).dt.date
oil_df = oil_df[['date', 'price_usd']]
oil_df = oil_df.where(pd.notnull(oil_df), None)

sql = "INSERT IGNORE INTO Oil_prices (date, price_usd) VALUES (%s, %s)"

data_list = [tuple(x) for x in oil_df.to_numpy()]
for row in data_list:
    cursor.execute(sql, row)

connection.commit()
print(f"   Done! ({len(data_list)} rows processed)")


# ---------------------------------------------------------
# 4. UPLOAD STOCK PRICES
# ---------------------------------------------------------
print("4. Uploading Stock Prices (Row by Row)...")

stocks_clean = stocks_df.copy()

# Rename columns
stocks_clean = stocks_clean.rename(columns={
    'Date': 'date',
    'tickers': 'ticker',
    'Open': 'open',
    'High': 'high',
    'Low': 'low',
    'Close': 'close',
    'Volume': 'volume'
})

stocks_clean['date'] = pd.to_datetime(stocks_clean['date']).dt.date
final_stocks = stocks_clean[['date', 'open', 'high', 'low', 'close', 'volume', 'ticker']]
final_stocks = final_stocks.where(pd.notnull(final_stocks), None)

sql = """
INSERT IGNORE INTO Stock_prices (date, open, high, low, close, volume, ticker) 
VALUES (%s, %s, %s, %s, %s, %s, %s)
"""

data_list = [tuple(x) for x in final_stocks.to_numpy()]
for row in data_list:
    cursor.execute(sql, row)

connection.commit()
print(f"   Done! ({len(data_list)} rows processed)")

# Close connection
connection.close()
print("\n🎉 All uploads finished successfully!")
```
